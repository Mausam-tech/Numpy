```python
# Q4 — Exam Score Slicing (1D indexing & slicing)
# The scores of 20 students, in roll-number order:

scores = np.array([45, 78, 88, 56, 90, 62, 71, 84, 39, 95,
                   67, 73, 81, 58, 92, 49, 77, 85, 63, 70])
# Using only indexing/slicing (no loops, no conditions):
# The score of the first and the last student.
print(scores[[0,-1]])
# The scores of roll numbers 5 to 10 (inclusive).
print(scores[5:10])
# Every 3rd student's score, starting from roll number 1.
print(scores[0::3])
# The scores in reverse order.
print(scores[::-1])
# The five lowest-roll-numbered students' scores, excluding the very first student.
print(scores[1:5])



```

```python
# # Q5 — Daily Temperature Report (basic statistics)
# # Hourly temperatures (°C) in Kathmandu for one day:

np.random.seed(15)
temps = np.round(np.random.uniform(8, 27, 24), 1)
# Report the maximum, minimum, mean and standard deviation of the day.
print(f"maximun temperature will be {temps.max()}degree celsious ")
print(f"minimum temperature will be {temps.min()} degree celcious" )
print(f"average temperature will be {temps.mean()} degree celcious" )
print(temps.std())

# Report the median and explain in a comment when the median is a better summary than the mean.
# if data have outlier median is better in those cases. 
# Find at which hour (index) the max and min occurred (argmax / argmin).
print(f"Max temperature at hour index {temps.argmax()}") 
print(f"Min temperature at hour index {temps.argmin()}")
# The forecast said the average would be 18 °C — print by how many degrees the real mean missed the forecast.
forecast = 18
miss = temps.mean() - forecast
print(f"The real mean missed the forecast by {miss:.1f} °C")

```
