```python
# Q2 — Cinema Seating Plan (zeros, ones, full)
# A new cinema hall has 12 rows and 16 seats per row.
# Create the empty hall as an array of zeros (0 = free seat).
hall = np.zeros((12,16))
print(hall)
# A private screening books the entire hall — create the same-shaped array filled with ones (1 = booked).
booked_hall = np.ones((12,16))
print(booked_hall)
# Ticket price is Rs. 450 for every seat — create a price grid of shape (12, 16) filled with 450 using np.full.
hall_cost = np.full((12,16),450)
print(hall_cost)
# Print the total revenue if the hall sells out (use the price grid, not arithmetic by hand).
print(f"Total revenue of the hall is Rs.{np.sum(hall_cost)}")
```

```python
# Q3 — Bus Departure Timetable (arange & linspace)
# A bus park manages departures from 6:00 (hour 6.0) to 21:00 (hour 21.0).

# Route A departs every 45 minutes starting at 6:00 — generate the departure times in hours 

# using np.arange (45 minutes = 0.75 hours). The last bus must not depart after 21:00.
route_a = np.arange (6,21, 0.75)
print(f" Depature time of Bus is: {route_a} ")
# Route B makes exactly 12 equally spaced departures between 6:00 and 21:00 inclusive — generate them with np.linspace.
# Print how many departures each route makes per day.
route_B = np.linspace(6 ,21,12)
print(route_B)
# In a comment, explain which function you would pick 
# when the interval is fixed vs when the number of trips is fixed.
i will use linspace when the interval and number of trips is fixed because 
linespace is efficent for it. 
```

```python
# Q4 — Exam Score Slicing (1D indexing & slicing)
# The scores of 20 students, in roll-number order:

scores = np.array([45, 78, 88, 56, 90, 62, 71, 84, 39, 95,
                   67, 73, 81, 58, 92, 49, 77, 85, 63, 70])
# Using only indexing/slicing (no loops, no conditions):
# The score of the first and the last student.
print(scores[[0,-1]])
# The scores of roll numbers 5 to 10 (inclusive).
print(scores[5:10])
# Every 3rd student's score, starting from roll number 1.
print(scores[0::3])
# The scores in reverse order.
print(scores[::-1])
# The five lowest-roll-numbered students' scores, excluding the very first student.
print(scores[1:5])



```
